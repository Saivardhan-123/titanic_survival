```python
df['Sex'] = df['Sex'].map({'male':0,'female':1})
```

```python
df['Embarked'] = df['Embarked'].map({'S':1,'C':2,'Q':3,'nan':4})
```

```python
df.dropna(subset=['Survived'], inplace=True)
```

```python
df = df.drop('Embarked',axis=1)
```

```python
X = df.drop('Survived',axis=1)
y = df['Survived']
```

```python
X
```

```python
y
```

```python
df.dtypes
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X_train,y_train)
```

```python
X_train_pred = model.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score

acc = accuracy_score(X_train_pred,y_train)

print('The accuracy is ',acc)
```

```python
X_test_pred = model.predict(X_test)
```

```python
acc = accuracy_score(X_test_pred,y_test)

print('The accuracy is ',acc*100)
```

```python
data = np.asarray(['1','0','16','1.4','2','45.5'],dtype=float).reshape(1,-1)
```

```python
res = model.predict(data)
```
